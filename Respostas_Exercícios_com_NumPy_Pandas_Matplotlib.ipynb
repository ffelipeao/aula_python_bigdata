# Exercício 1 — Análise de Array NumPy

Você recebeu um array NumPy com os valores de temperatura (em °C) registrados ao longo de uma semana:



1.1.   Calcule a temperatura média da semana.

1.2.   Encontre a temperatura mais alta e a mais baixa.

1.3.   Quantos dias ficaram acima da média?

```python
import numpy as np
temperaturas = np.array([22.5, 23.0, 19.5, 21.0, 25.2, 26.3, 24.5])
```


**Solução do Exercício 1**

1.1- Calcule a temperatura média da semana

```python
media = temperaturas.mean()
print(media)

print('Arredondando com duas casas decimais apenas:')
print(round(media, 2))
```

1.2 - Encontre a temperatura mais alta e a mais baixa.

```python

max_temp = temperaturas.max()
min_temp = temperaturas.min()

print(f"Máxima: {max_temp}°C, Mínima: {min_temp}°C")

print("Máxima:", max_temp, "°C, Mínima: ", min_temp, "°C")

```

1.3.   Quantos dias ficaram acima da média?


```python
media = temperaturas.mean()
dias_acima_media = np.sum(temperaturas > media)


print(f"Média: {media:.2f} °C")
print("Média:", round(media, 2), "°C")


print(f"Dias acima da média: {dias_acima_media}")

```

Exercício 5 — Salvando Aprovados em um Arquivo .txt

1. Crie dados de 10 alunos e suas notas em **três avaliações** da disciplina de Programação Python.

2. Calcule a **média das três notas** para cada aluno.

3. Filtre os alunos com **média maior que 8**.

4. Salve os nomes dos aprovados em um arquivo chamado aprovados.txt, um nome por linha. Ao lado do nome coloca a seguinte mensagem: "Parabéns, você foi aprovado em Python."

5. Leia o arquivo aprovados.txt e mostre os nomes dos aprovados com a mensagem.


**Solução do Exercício 5**

```python
import pandas as pd

# 1. Criando os dados
dados = {
    'nome': ['Ana', 'Bruno', 'Carla', 'Diego', 'Eva', 'Felipe', 'Giulia', 'Heitor', 'Isabela', 'João'],
    'nota1': [9.0, 7.5, 8.5, 6.0, 9.5, 8.2, 7.8, 8.8, 9.1, 7.0],
    'nota2': [8.5, 6.5, 9.0, 7.0, 8.5, 8.0, 8.2, 9.0, 9.5, 7.5],
    'nota3': [9.0, 8.0, 8.5, 7.0, 9.0, 9.0, 7.5, 8.7, 8.8, 6.5]
}

df = pd.DataFrame(dados)

# 2. Calculando média
df['media'] = df[['nota1', 'nota2', 'nota3']].mean(axis=1)

# 3. Filtrando aprovados
aprovados = df[df['media'] > 8]

# 4. Salvando nomes com mensagem personalizada
mensagem = "Parabéns, você foi aprovado em Python."
with open('aprovados.txt', 'w') as f:
    for nome in aprovados['nome']:
        f.write(f"{nome} - {mensagem}\n")

# 5. Lendo e mostrando os aprovados com mensagem
with open('aprovados.txt', 'r') as f:
    print("Mensagens para aprovados:")
    for linha in f:
        print(linha.strip())

```

## Exercício 6 — Nuvem de Palavras com Notícias

**Objetivo**: Gerar uma nuvem de palavras a partir do conteúdo de uma página de notícias online para identificar os termos mais frequentes e, consequentemente, os temas mais discutidos do dia.

### Tarefas:

1.  **Escolha uma Página de Notícias**: Selecione uma URL de uma página de notícias (por exemplo, um portal de notícias, blog, etc.).
2.  **Extraia o Texto**: Utilize uma biblioteca (como `requests` para obter o HTML e `BeautifulSoup` para parsear) para extrair o texto principal da notícia ou de múltiplos artigos da página.
3.  **Pré-processamento do Texto**: Limpe o texto removendo caracteres especiais, números, pontuação e stopwords (palavras comuns como 'de', 'a', 'o', 'e', 'para', etc.). Converta todas as palavras para minúsculas.
4.  **Gere a Nuvem de Palavras**: Utilize a biblioteca `wordcloud` para criar uma nuvem de palavras a partir do texto pré-processado.
5.  **Exiba a Nuvem de Palavras**: Mostre a imagem da nuvem de palavras usando `matplotlib`.

**Solução do Exercício 6**

```python
import requests
from bs4 import BeautifulSoup
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

# 1. Escolha uma Página de Notícias
# Exemplo: Você pode mudar esta URL para a notícia que desejar
url = 'https://g1.globo.com/' # Exemplo de URL

# 2. Extraia o Texto
try:
    response = requests.get(url)
    response.raise_for_status() # Lança um erro para status codes HTTP ruins
    soup = BeautifulSoup(response.text, 'html.parser')

    # Tenta encontrar o texto em parágrafos ou em tags de artigos
    paragraphs = soup.find_all('p')
    text = ' '.join([p.get_text() for p in paragraphs])

    # Se não encontrar muito texto em 'p', tenta extrair de outras tags comuns em notícias
    if len(text.split()) < 100: # Se o texto for muito curto, tenta outra estratégia
        articles = soup.find_all('article')
        text = ' '.join([a.get_text() for a in articles])
        if len(text.split()) < 100: # Último recurso: todo o corpo da página
            text = soup.body.get_text()

    if not text:
        print("Não foi possível extrair texto suficiente da página.")

except requests.exceptions.RequestException as e:
    print(f"Erro ao acessar a URL: {e}")
    text = ""

# 3. Pré-processamento do Texto
# Limpeza básica e remoção de stopwords em português

# Adicione stopwords em português
stopwords_pt = set(STOPWORDS)
stopwords_pt.update(['de', 'a', 'o', 'que', 'do', 'da', 'em', 'um', 'para', 'é', 'com', 'não', 'uma', 'os', 'no', 'se', 'na', 'por', 'mais', 'as', 'dos', 'como', 'mas', 'foi', 'ao', 'ele', 'das', 'tem', 'à', 'seu', 'sua', 'ou', 'ser', 'quando', 'muito', 'há', 'nos', 'já', 'está', 'eu', 'também', 'nós', 'você', 'vocês', 'eles', 'elas', 'este', 'isto', 'isto', 'isso', 'esse', 'aquilo', 'aqui', 'lá', 'onde', 'quem', 'meu', 'minha', 'meus', 'minhas', 'teu', 'tua', 'teus', 'tuas', 'nosso', 'nossa', 'nossos', 'nossas', 'vosso', 'vossa', 'vossos', 'vossas', 'isso', 'isto', 'e', 'após', 'entre', 'todos', 'tudo', 'disse', 'apenas', 'ainda', 'depois', 'agora', 'vez', 'sem', 'antes', 'pode', 'onde', 'porque', 'onde', 'mesmo', 'dias', 'pelo', 'vezes'])

# Remove pontuação, números e converte para minúsculas
clean_text = ''.join([char for char in text if char.isalpha() or char.isspace()])
clean_text = clean_text.lower()

# Remove stopwords
words = clean_text.split()
filtered_words = [word for word in words if word not in stopwords_pt and len(word) > 2] # Remove palavras de 1 ou 2 letras
processed_text = ' '.join(filtered_words)

# 4. Gere a Nuvem de Palavras
if processed_text:
    wordcloud = WordCloud(width = 800, height = 400,
                          background_color ='white',
                          stopwords = stopwords_pt,
                          min_font_size = 10).generate(processed_text)

    # 5. Exiba a Nuvem de Palavras
    plt.figure(figsize = (10, 5), facecolor = None)
    plt.imshow(wordcloud)
    plt.axis("off")
    plt.tight_layout(pad = 0)
    plt.show()
else:
    print("Não há texto processado para gerar a nuvem de palavras.")
```
