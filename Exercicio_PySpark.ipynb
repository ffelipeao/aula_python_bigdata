## 2. Configuração das Variáveis de Ambiente

Precisamos informar ao sistema onde o Java e o Spark estão instalados para que o PySpark possa encontrá-los corretamente.

```python
# Configurar as variáveis de ambiente para Java e Spark
import os
import findspark

os.environ["JAVA_HOME"] = "/usr/lib/jvm/java-8-openjdk-amd64"
os.environ["SPARK_HOME"] = "/usr/local/lib/python3.12/dist-packages/pyspark" # Força o SPARK_HOME para o diretório correto
findspark.init()

print("Variáveis de ambiente configuradas!")
```

## 3. Inicialização do SparkSession

**SparkSession** é o ponto de entrada para programar o Spark com a API DataFrame e Dataset. Ele é a forma mais comum de interagir com o Spark e criar DataFrames.

```python
# Importar e inicializar o SparkSession
from pyspark.sql import SparkSession

# Criar uma SparkSession
# .builder cria um construtor para a SparkSession
# .appName define um nome para a sua aplicação Spark
# .getOrCreate() obtém uma SparkSession existente ou cria uma nova se não houver
spark = SparkSession.builder.appName("PySparkColabClassroom").getOrCreate()

print("SparkSession inicializada com sucesso!")

# Exibir a versão do Spark e informações da sessão
print(spark.version)
spark
```

## 4. Análise do Dataset `california_housing_test.csv` com Pandas e PySpark

O dataset `california_housing_test.csv` contém dados demográficos e de habitação para distritos da Califórnia, utilizados para prever valores medianos de casas. É uma versão de teste do famoso dataset California Housing. Cada linha representa um grupo de blocos.

### Colunas Principais:

*   **`longitude`**: Uma medida de quão longe a casa está a oeste (um valor mais alto é mais a oeste).
*   **`latitude`**: Uma medida de quão longe a casa está ao norte (um valor mais alto é mais ao norte).
*   **`housing_median_age`**: Idade mediana da casa em um bloco.
*   **`total_rooms`**: Número total de quartos dentro do bloco.
*   **`total_bedrooms`**: Número total de quartos de dormir dentro do bloco.
*   **`population`**: População total dentro do bloco.
*   **`households`**: Total de domicílios (famílias), um grupo de indivíduos que residem em uma unidade habitacional.
*   **`median_income`**: Renda mediana para os domicílios dentro do bloco (em dezenas de milhares de dólares).
*   **`median_house_value`**: Valor mediano da casa para os domicílios dentro do bloco (o que a previsão tenta atingir).
