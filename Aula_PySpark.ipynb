## Introdução ao PySpark no Google Colab

### O que é PySpark?

**Apache Spark** é um mecanismo de análise unificado e de código aberto para processamento de dados em larga escala. Ele é conhecido por sua velocidade, facilidade de uso, capacidade de processamento analítico avançado e flexibilidade.

**PySpark** é a API Python para o Apache Spark. Ele permite que cientistas de dados e engenheiros de dados que trabalham com Python aproveitem o poder do Spark para analisar grandes conjuntos de dados. Com PySpark, você pode escrever aplicativos Spark usando a linguagem Python, que é muito popular para análise de dados e aprendizado de máquina.

### Por que usar PySpark no Colab?

*   **Processamento de Big Data**: PySpark é excelente para lidar com datasets que não cabem na memória de uma única máquina.
*   **Escalabilidade**: Permite escalar o processamento de dados horizontalmente, utilizando clusters de máquinas.
*   **Velocidade**: Oferece performance superior em comparação com o Pandas para grandes volumes de dados, graças à sua arquitetura distribuída e otimizações.
*   **Ecossistema Rico**: Integra-se com várias bibliotecas de machine learning (MLlib) e ferramentas de streaming.
*   **Gratuito e Fácil Acesso**: O Google Colab oferece um ambiente gratuito e baseado em nuvem para executar PySpark sem a necessidade de configurações complexas de infraestrutura.

## 5. Exemplo Prático: Análise de Vendas (Simulação de Dados)

**Cenário**: Uma pequena loja registra suas vendas diárias de alguns produtos.

### Exemplo 5.1: Carregando dados de um arquivo CSV e Análise Inicial

Primeiro, vamos criar um arquivo CSV de exemplo e carregá-lo no PySpark.

```python
# Criar um arquivo CSV de exemplo com mais dados
%%writefile vendas.csv
Data,Produto,Quantidade,PrecoUnitario
2023-01-01,Maçã,10,2.50
2023-01-01,Banana,15,1.20
2023-01-01,Laranja,8,3.00
2023-01-01,Uva,5,4.00
2023-01-02,Maçã,12,2.50
2023-01-02,Banana,20,1.20
2023-01-02,Abacaxi,5,5.00
2023-01-02,Pêra,7,3.50
2023-01-03,Laranja,10,3.00
2023-01-03,Maçã,7,2.50
2023-01-03,Banana,18,1.20
2023-01-03,Morango,12,6.00
2023-01-04,Maçã,8,2.50
2023-01-04,Laranja,15,3.00
2023-01-04,Uva,10,4.00
2023-01-04,Abacaxi,6,5.00
2023-01-05,Banana,25,1.20
2023-01-05,Morango,10,6.00
2023-01-05,Pêra,9,3.50
2023-01-05,Maçã,11,2.50
2023-01-06,Laranja,12,3.00
2023-01-06,Uva,8,4.00
2023-01-06,Banana,14,1.20
2023-01-06,Abacaxi,7,5.00
2023-01-07,Maçã,10,2.50
2023-01-07,Morango,15,6.00
2023-01-07,Pêra,10,3.50

print("Arquivo 'vendas.csv' criado com mais dados!")
```

```python
# Carregar o arquivo CSV em um DataFrame Spark
# header=True indica que a primeira linha contém os nomes das colunas
# inferSchema=True tenta adivinhar o tipo de dado de cada coluna
df_vendas = spark.read.csv("vendas.csv", header=True, inferSchema=True)

print("DataFrame 'df_vendas' carregado!")

print("Primeiras linhas do DataFrame de vendas:")
df_vendas.show()
```

```python
print("Estatísticas de uma coluna:")
df_vendas.describe("Quantidade").show()
```

### Exemplo 5.2: Calculando o Total de Vendas por Produto

Vamos calcular uma nova coluna `TotalVenda` e depois agrupar os dados por `Produto` para somar as vendas.

```python
from pyspark.sql.functions import col, sum, round

# 1. Calcular a coluna 'TotalVenda' (Quantidade * PrecoUnitario)
# withColumn adiciona uma nova coluna ou substitui uma existente
df_vendas_com_total = df_vendas.withColumn("TotalVenda", round(col("Quantidade") * col("PrecoUnitario"),2))

print("DataFrame com coluna 'TotalVenda':")
df_vendas_com_total.show()
```

```python
# 2. Agrupar por 'Produto' e somar o 'TotalVenda'

# groupBy define as colunas para agrupar
# agg realiza agregações (aqui, soma do TotalVenda)

df_total_por_produto = df_vendas_com_total.groupBy("Produto").agg(sum("TotalVenda").alias("SomaTotalVendas"))

print("Total de Vendas por Produto:")
df_total_por_produto.show()
```

### Exemplo 5.3: Consultando Dados de Vendas com Spark SQL

Podemos usar Spark SQL para realizar consultas em DataFrames como se fossem tabelas de um banco de dados. Isso nos permite usar a sintaxe SQL que muitos analistas e engenheiros já estão acostumados.

```python
# 1. Registrar o DataFrame 'df_vendas_com_total' como uma view temporária SQL
df_vendas_com_total.createOrReplaceTempView("vendas_sql")

print("DataFrame 'df_vendas_com_total' registrado como a view temporária 'vendas_sql'.")
```

```python
# 2. Executar uma consulta SQL para selecionar todas as vendas
print("Todas as vendas via Spark SQL:")
spark.sql("SELECT * FROM vendas_sql").show()
```

```python
# 3. Executar uma consulta SQL para encontrar as vendas totais por produto
print("Total de Vendas por Produto via Spark SQL:")
spark.sql("SELECT Produto, SUM(TotalVenda) AS TotalVendasSQL FROM vendas_sql GROUP BY Produto ORDER BY TotalVendasSQL DESC").show()
```

```python
# 4. Executar uma consulta SQL para filtrar vendas acima de um determinado valor
print("Vendas individuais com TotalVenda maior que 30.00 via Spark SQL:")
spark.sql("SELECT Data, Produto, TotalVenda FROM vendas_sql WHERE TotalVenda > 30.00 ORDER BY TotalVenda DESC").show()
```
